```python
from google.colab import files
import pandas as pd
import io

uploaded = files.upload()

if not uploaded:
    print("Ошибка: Вы не выбрали файл для загрузки.")
else:
    # Берём первый загруженный файл
    filename = next(iter(uploaded))

    df = pd.read_csv(io.BytesIO(uploaded[filename]))

    print("Файл:", filename)
    print("Датасет успешно загружен!")
    display(df.head())
```

```python

# Размер таблицы
print("Количество строк:", df.shape[0])
print("Количество столбцов:", df.shape[1])

# Названия и типы столбцов
print("\nТипы данных:")
print(df.dtypes)

# Число пропусков в каждом столбце
print("\nПропуски по столбцам:")
print(df.isnull().sum())

# Процент пропусков
print("\nДоля пропусков (%):")
print((df.isnull().mean() * 100).round(2))

# Базовые статистики числовых столбцов
print("\nСтатистики числовых признаков:")
display(df.describe())

# Статистики всех столбцов, включая категориальные
print("\nСтатистики всех признаков:")
display(df.describe(include="all"))
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

numeric_cols = df.select_dtypes(include="number").columns.tolist()

if numeric_cols:
    col = numeric_cols[0]
    values = df[col].dropna()

    plt.figure(figsize=(8, 5))
    sns.histplot(values, bins=20, kde=True)
    plt.title(f"Распределение признака: {col}")
    plt.xlabel(col)
    plt.ylabel("Частота")
    plt.tight_layout()
    plt.show()
else:
    print("В датасете нет числовых столбцов.")
```
